**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class student:

  def __init__(self, name, roll_number):
    self.__name = name  # Private variable
    self.__roll_number = roll_number  # Private variable

  # Getter methods
  def get_name(self):
    return self.__name

  def get_roll_number(self):
    return self.__roll_number

  # Setter methods
  def set_name(self, name):
    self.__name = name

  def set_roll_number(self, roll_number):
    self.__roll_number = roll_number


s1 = student("Ali", 450)
print(s1.get_name(), s1.get_roll_number())

s1.set_name("Hamza")
s1.set_roll_number(500)
print(s1.get_name(), s1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:

  def __init__(self, initial_balance=0):
    self.__balance = initial_balance  # Private variable

  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
      print(f"{amount} deposit ho gaye. New Balance: {self.__balance}")
    else:
      print("Invalid deposit amount!")

  def withdraw(self, amount):
    if amount <= 0:
      print("Invalid withdrawal amount!")
    elif amount > self.__balance:
      print(
          "Transaction Failed: Balance zero ya insufficient nahi ho sakta!"
      )  # Check condition
    else:
      self.__balance -= amount
      print(f"{amount} nikal liye gaye. New Balance: {self.__balance}")

  def get_balance(self):
    return self.__balance


account = BankAccount(5000)
account.deposit(2000)
account.withdraw(3000)
account.withdraw(6000)  # Yeh check karega ke balance se zyada na nikle
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def __init__(self):
        self.living = True
    def eat(self):
        print("animal can eat")
        
class Mammal(Animal):
    def __init__(self):
        super().__init__()
        self.warm_blooded = True
        
    def walk(self):
        print("Mmammal can walk")

class Dog (Mammal):
    def __init__(self):
        super().__init__()
        self.breed = "Dogo Argentino"
    def voice(self):
        print("Dog is barking")        
```

```python
d = Dog()
print(d.living, d.warm_blooded, d.breed )
d.eat()
d.walk()
d.voice()
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
# Write your code here for Q9
class Person:

  def __init__(self, name):
    self.name = name
    print("Parent constructor called!")


class Student(Person):

  def __init__(self, name, grade):
    # super() ke zariye Parent class ke constructor ko call kiya
    super().__init__(name)
    self.grade = grade
    print("Child constructor called!")


s = Student("Ali", "A+")

print(f"Name: {s.name}")
print(f"Grade: {s.grade}")
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class Shape:
    def info(self):
        print("its a shape")

class Circle(Shape):
    def area(self, r):
        return 3.14 * r * r

class Square(Shape):
    def area(self, side):
        return side * side

c = Circle()
s = Square()
c.info()
print("Circle Area:", c.area(44))
print("Square Area:", s.area(35))
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(234, 567)
p2 = Point(200,700)
p3 = p1 + p2
print(p3)
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
# Write your code here for Q19
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCard(Payment):
    def pay(self, amount):
        print(f"{amount} paid using Credit Card")

class MeezanBank (Payment):
    def pay(self, amount):
        print(f"{amount} paid using Meezan Bank")

class Wallet:
    def __init__(self):
        self.__balance = 100000

    def get_balance(self):
        return self.__balance

    def add_money(self, amount):
        self.__balance += amount
        print(f"{amount} added")

print("--- Abstracration Testing ---")
p1 = CreditCard()
p1.pay(50000)
      
p2 = MeezanBank()
p2.pay(20000)

print("\n=== Encapsulation Testing ---")
w = Wallet()
print(f"Current Balance: {w.get_balance()}")  
w.add_money(5000)
print(f"New Balance: {w.get_balance()}")
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    def payment_successful(self, amount):
        print(f"Payment of {amount} Rs successful")

class CreditCardPayment(Payment):        
    def __init__(self, card_number):
        self.card_number = card_number

    def pay(self, amount):
        print(f"Paying {amount} Rs using Credit Card ending with {self.card_number[-4:]}")

class UPIPayment(Payment):
    def __init__(self, upi_id):
        self.upi_id = upi_id
         
    def pay(self, amount):
        print(f"{amount} paid using UPI ID {self.upi_id}")

class Wallet:
    def __init__(self):
        self.__balance = 100

    def get_balance(self):
        return self.__balance
    
credit = CreditCardPayment("4321-8765-0921-6543")
credit.pay(5100)
upi = UPIPayment("Faiq@MeezanBank")
upi.pay(2100)

w = Wallet()
print(f"Wallet Balance: {w.get_balance()}")
       
```
