```python
# Q1 - EVEN NUMBERS 1-20
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

# Q2 - ODD NUMBERS 1-20

```python
for i in range(1, 21):
    if i % 2 != 0:
        print(i)
```

# Q3 - 1-10, 5 special print

```python
for i in range(1, 11):
    if i == 5:
        print("Five found!")
    else:
        print(i)
```

# Q4 - LIST [3,8,12,1,9],>5 print

```python
numbers = [3, 8, 12, 1, 9]
for n in numbers:
    if n > 5:
        print(n)
```

# Q5 - 1-15, DIVISIBLE BY 3

```python
for i in range(1, 16):
    if i % 3 == 0:
        print(i)
```

# Q6 - [2,-5,10,-8,0] POSITIVE/NEGATIVE/ZERO

```python
nums = [2, -5, 10, -8, 0]
for n in nums:
    if n > 0:
        print("Positive")
    elif n < 0:
        print("Negative")
    else:
        print("Zero")
```

# Q7 - 1-10, BREAK ON 7

# Q8 - 1-10, skip 4

```python
for i in range(1, 11):
    if i == 4:
        continue
    print(i)
```

# Q9 - VOWELS IN 'PYTHON'

# Q10 - AGES ADULT/MINOR

```python
ages = [12, 18, 25, 15]
for a in ages:
    if a >= 18:
        print("Adult")
    else:
        print("Minor")
```

# Q11 - NAMES, SARA CHECK

# Q13 - EVEN 2, else same

# Q14 - NAMES STARTING WITH 'A'

```python
names = ["Ali", "Bilal", "Ahmed", "Sana"]
for n in names:
    if n.startswith("A"):
        print(n)
```

# Q15 - COUNT 'apple'

```python
count = 0
fruits = ["apple", "mango", "apple", "banana", "apple"]
for f in fruits:
    if f == "apple":
        count += 1
print(count)
```

# Q17 - NUMBERS BETWEEN 100 - 200

# Q18 - REVERSE 10->1, HALFWAY

```python
for i in range(10, 0, -1):
    if i == 5:
        print("Halfway there!")
    print(i)
```

# Q19 - SUM OF EVEN 1-10

```python
total = 0
for i in range(1, 11):
    if i % 2 == 0:
        total += i
print(total)
```
